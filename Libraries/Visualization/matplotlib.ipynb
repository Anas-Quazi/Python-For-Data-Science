```python

#! twinx: two y-axes sharing one x-axis
epochs = np.arange(1, 11)
cpu = [45, 52, 58, 63, 65, 68, 70, 71, 72, 72]
loss = [2.1, 1.6, 1.2, 0.9, 0.75, 0.6, 0.5, 0.45, 0.4, 0.38]

fig, ax1 = plt.subplots(figsize=(9, 5))
ax1.plot(epochs, cpu, color="#F5420C", marker='o', label='CPU temp')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Temp (°C)', color="#F5420C")
ax1.tick_params(axis='y', colors="#F5420C")

ax2 = ax1.twinx()
ax2.plot(epochs, loss, color="#0CF5ED", marker='s', label='Loss')
ax2.set_ylabel('Loss', color="#0CF5ED")
ax2.tick_params(axis='y', colors="#0CF5ED")

#^ one combined legend
h1, l1 = ax1.get_legend_handles_labels()
h2, l2 = ax2.get_legend_handles_labels()
ax1.legend(h1 + h2, l1 + l2, loc='center right')
plt.show()
```
