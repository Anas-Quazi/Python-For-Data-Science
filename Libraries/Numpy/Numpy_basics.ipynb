## Array reshaping and resizing

## Indexing & Slicing

```python

#^ indexing of vector

vector1 = np.arange(16, 26)

print(vector1)
print('2nd element : ', vector1[1])
print('3rd last : ', vector1[-3])
print()
```

```python

#^ slicing array

print(vector1[1:4])
print(vector1[2:])
print(vector1[:3])
print(vector1[0:6:3])

#~ -ve slicing
print(vector1[-4:-1])
print(vector1[-4:-1:2])

#& step
print(vector1[::3])
```

```python

#? indexing and slicing of a matrix
matrix1 = np.arange(1, 17).reshape(4, 4)
print(matrix1)

print('\n1st row : ', matrix1[0])
print('\n1st element: ', matrix1[0,0])
print(matrix1[2,3])

#^ slicing a matrix
print('\nsliced : \n', matrix1[0:2, 1:5])
print('\n', matrix1[2:, 2:])

print('\nanother way : ', matrix1[0][2])
print('\ncolumn 3 : ', matrix1[:,2])
```

```python

#* boolean indexing
arr1 = np.arange(1, 11)
print(arr1)

bool_idx = arr1 % 2 == 0
print(bool_idx)

#todo applying boolean index to existing array
arr1 = arr1[bool_idx]
print(arr1)
```

## Deep and shallow copy

```python

#& shallow copy
a = np.arange(1, 31)
print(a)

slc = a[:10]
slc = slc * 2
print(slc)

print('\n no changes due to shallow copy', a)
```

```python

#? deep copy 
b = np.arange(1, 6)
print(b)
c = b #~ use c = b.copy for shallow copy
c[0] = 12
print(b)
```

## vector & matrix operations

```python

vec1 = np.array([1, 2, 3, 4])
vec2 = np.array([10, 15, 20, 25])

print('dot product : ', vec1 @ vec2)
print('cross product : ', np.cross(vec1, vec2))
```

```python

#todo .vectorize() function

print(vec2)

def classify(x):
    return 'even' if x % 2 == 0 else 'odd'

vec_classify = np.vectorize(classify)

print(vec_classify(vec2))
```

```python

#! valid sudoku
def validSudoku(grid): #~ wrong one...

    rows_sum = grid.sum(axis=1)
    colm_sum = grid.sum(axis=0)

    valid = True
    for i in rows_sum:

        if i != 45:
            valid = False

    if valid:
        for i in colm_sum:
            if i != 45:
                valid = False
    if valid:
        print('sudoku is valid!')
    else:
        print('invalid...')

def validSudoku_corrected(grid):

    #todo check for unique rows and columns
    for i in range(9):

        if len(np.unique(grid[i, :])) != 9:
            return False
        if len(np.unique(grid[:, i])) != 9:
            return False

    #? check for valid 3x3 subgrid
    for r in range(0, 9, 3):
        for c in range(0, 9, 3):
            subgrid = grid[r:r+3, c:c+3]
            if len(np.unique(subgrid)) != 9:
                return False

    return True

sudoku_array = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 4, 2, 1, 7, 6, 3],
    [4, 2, 6, 8, 5, 7, 9, 1, 3],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])

valid_grid = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])

validSudoku_corrected(sudoku_array)
validSudoku_corrected(valid_grid)
```
